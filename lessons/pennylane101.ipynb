**Introduction to Pennylane**

```python
import pennylane as qp
```

```python
def circuit():
    qp.X(wires=0)
    return qp.probs()

qp.draw_mpl(circuit)()
```

```python
def circuit2():
    qp.CNOT(wires=[0,1])
    return qp.probs()

qp.draw_mpl(circuit2)()
```

```python
def circuit3(x,y):
    qp.RX(x,wires=0)
    qp.RY(y,wires=0)
    return qp.probs()

qp.draw_mpl(circuit3, decimals=1)(0.1,0.2)
```

```python
dev = qp.device("default.qubit")

@qp.qnode(dev)
def circuit3(x,y):
    qp.RX(x,wires=0)
    qp.RY(y,wires=0)
    return qp.probs()

qnode1 = qp.QNode(circuit2,dev)

# circuit3(0.1,0.2)
qnode1()
```
